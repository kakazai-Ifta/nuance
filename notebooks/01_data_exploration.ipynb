```python
review_files = glob.glob("../data/raw/reviews*.csv")
dataframes = []
for file in review_files:
    # Read CSV
    df = pd.read_csv(file)
    # Add to datafram list
    dataframes.append(df)
    
    df_reviews = pd.concat(dataframes, axis=0, ignore_index=True)
```

```python
# Reading in the product_info.csv file into a dataframe
df_products = pd.read_csv("../data/raw/product_info.csv")
```

```python
# Checking for duplicate product_id values in df_products
df_products["product_id"].duplicated().sum()
```

```python
# Creating a subset of df_products with only the necessary columns for merging with df_reviews
df_products_subset = df_products[["product_id",
                                 "primary_category",
                                 "secondary_category",
                                 "tertiary_category",
                                 "size",
                                 "loves_count"]]
df_merged = pd.merge(df_reviews, df_products_subset, on="product_id", how="left", validate="many_to_one")
```

```python
# Data counts and value counts for the merged dataframe
print(df_merged.shape)
print(df_merged["primary_category"].value_counts(dropna=False))
print(df_merged["secondary_category"].value_counts(dropna=False))
print(df_merged["tertiary_category"].value_counts(dropna=False))
```

```python
#Exporting the merged dataframe to parquet format for further analysis
df_merged.to_parquet("../data/processed/df_merged.parquet", index=False)
```
